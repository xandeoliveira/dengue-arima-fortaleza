```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.dates import DateFormatter, MonthLocator
```

```python
sarima = pd.read_csv("dataset/previsoes_sarima.csv")
lstm = pd.read_csv("dataset/previsoes_lstm.csv")
```

```python
df = pd.DataFrame({
  'data': lstm['Data'],
  'lstm': lstm['Previsão'],
  'sarima': sarima['Point.Forecast'],
  'ich95': sarima['Hi.95'],
  'icl95': sarima['Lo.95']
})
```

```python
df['data'] = pd.to_datetime(df['data'])
```

```python
#df.set_index('data')[['lstm', 'sarima']].plot(figsize=(10,5))

plt.figure(figsize=(16, 6))

plt.plot(df['data'], df['lstm'], label='LSTM', color='blue')
plt.plot(df['data'], df['sarima'], label='SARIMA', color='red')

ax = plt.gca()
ax.xaxis.set_major_formatter(DateFormatter('%b'))

plt.title("Comparação de previsão de casos de dengue no município de Fortaleza para o ano de 2025 - LSTM x SARIMA")
plt.xlabel("Meses")
plt.ylabel("Casos")

plt.grid(True)
plt.legend()
```
